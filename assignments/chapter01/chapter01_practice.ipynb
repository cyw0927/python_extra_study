```python
students = [
	{"name": "스티브로저스", "score": 91, "attendance": 95},
	{"name": "피터파커", "score": 60, "attendance": 78},
	{"name": "토르", "score": 70, "attendance": 70},
	{"name": "토니스타크", "score": 99, "attendance": 15}
]

for student in students:
	name = student["name"]
	score = student["score"]
	attendance = student["attendance"]
	
	if score < 60 or attendance < 60:
		grade = "F"
	elif score >= 90 and attendance >= 80:
		grade = "A"
	elif score >= 80 and attendance >= 70:
		grade = "B"
	elif score >= 70 and attendance >= 70:
		grade = "C"
	else:
		grade = "D"


	print( 
		f"{name}: 점수 {score}"
	    f" 출석률 {attendance}%"
		f" 성적 {grade}"
    )
```
