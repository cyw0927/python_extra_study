## 4. 데이터 준비

```python
import pandas as pd

url = "https://raw.githubusercontent.com/GilbertMoon/llm-data-analysis-course/main/data/raw/customers.csv"
df = pd.read_csv(url)

df.columns = [
    "Customer_ID",
    "Name",
    "Gender",
    "Age",
    "City",
    "Signup_Date"
]

records = df.to_dict(orient="records")
print("records type : ", type(records))
columns = list(df.columns)
```

```python
#print(records)
print(len(records))
print(records[0])
```

```python
print(df.info())
```

```python

max_age = 0
min_age = 1000

for rec in records:
    age = rec["Age"]

    if age > max_age: 
        max_age = age
    if age < min_age:
        min_age = age

print ("max_age : ", max_age)
print ("min_age : ", min_age)
        
```

```python
max_date = '1900-12-01'
min_date = '2999-12-01'

for rec in records:
    date = rec["Signup_Date"]

    if date > max_date: 
        max_date = date
    if date < min_date:
        min_date = date

print ("max_date : ", max_date)
print ("min_date : ", min_date)
```
