## 4. 데이터 준비

```python
import pandas as pd

url = "https://raw.githubusercontent.com/GilbertMoon/llm-data-analysis-course/main/data/raw/customers.csv"
df = pd.read_csv(url)

df.columns = [
    "Customer_ID",
    "Name",
    "Gender",
    "Age",
    "City",
    "Signup_Date"
]

records = df.to_dict(orient="records")
print("records type : ", type(records))
columns = list(df.columns)
```

```python
#print(records)
print(len(records))
print(records[0])
```

```python
print(df.info())
```

```python

max_age = 0
min_age = 1000

for rec in records:
    age = rec["Age"]

    if age > max_age: 
        max_age = age
    if age < min_age:
        min_age = age

print ("max_age : ", max_age)
print ("min_age : ", min_age)
        
```

```python
max_date = '1900-12-01'
min_date = '2999-12-01'

for rec in records:
    date = rec["Signup_Date"]

    if date > max_date: 
        max_date = date
    if date < min_date:
        min_date = date

print ("max_date : ", max_date)
print ("min_date : ", min_date)
```

## 5. 직접 구현
### 전체 고객 한 명씩 확인하기

### 날짜 최대/최소 구현 전 생각
- 반복할 값: 각 고객의 가입일
- 비교할 값: 현재까지 기억한 최근날짜 / 가장 오래된 날짜
- 반복 전에 준비할 변수: max date, min date
- 어떤 경우에 값을 바꿀 것인가: 현재 고객의 값이 최근 혹은 오래된 날짜에 해당되면 바꾼다
- 아직 헷갈리는 부분: 해당날짜에 해당되는 고객의 id는 어떻게 연동할것인가

### 나이 합계/평균 구현 전 생각

- 합계를 저장할 변수: 단순 합계이기 때문에 계속 더할 변수를 만든다
- 처음 값은 무엇으로 둘 것인가: 0
- 고객 한 명을 확인할 때 무엇을 더할 것인가: 고객의 나이
- 사람 수는 어떻게 셀 것인가: 고객 한명을 확인할때마다 별도의 값 1씩 추가한다
- 평균은 언제 계산할 것인가: 마지막 최종합계 계산 이후 /N으로 계산한다
- 아직 헷갈리는 부분: sum 안쓰고 이게 어떻게 돌아갈지 모르겠다


### 컬럼명 소문자 변환 전 생각
- 반복할 대상: columns List안에 컬럼 명
- 새 컬럼명을 저장할 자료구조: List
- 문자열 하나를 소문자로 바꾸는 방법: rename() or lower()
- 바뀐 문자열을 어디에 추가할 것인가: 새로 만든 List안에 append로 추가
- 아직 헷갈리는 부분: lower()이라는 코드가 있는지 모르겠다

### 무엇이 필요했는가?

### 처음에는 무엇을 몰랐는가?

lower_columns와 lower_column을 같은 변수처럼 써서
List가 문자열로 바뀌는 문제도 있었다


### 내가 먼저 생각한 내용

처음에 append를 빼놓고 치니 단순하게 list안 값만 바뀌었다

### 어디에서 확인했는가?
- 여기서 돌려보고 안되면 gpt에게 코드 쳐주지말고 방법을 물어봤다

### 확인 후 내가 이해한 내용

lower()는 문자열 하나를 소문자로 바꾸는 기능이고,
append()는 List에 값을 하나씩 추가하는 기능이다

lower_columns는 전체 결과를 저장하는 List이고,
lower_column은 현재 소문자로 바꾼 값 하나라는 차이가 있다

### 이 내용을 실제 코드의 어느 부분에 사용했는가?

바로 위 소문자로 바꾸는 코드에 사용했다


## 아직 이해가 안 되는 부분
### 항목 1
- 이해되지 않는 내용: 여전히 append 앞에 왜 .이 붙는지 완벽하게 이해하진 못했다
- 어디까지는 이해했는가: append()가 List에 값을 하나 추가하는 기능이라는 것은 이해했다
- 정확히 어느 부분부터 모르겠는가:왜 `append()`만 쓰는 게 아니라 `lower_columns.append()`처럼 앞에 변수명과 점(.)을 붙여야 하는지 아직 직관적으로 이해되지 않는다. 언제 .이 붙는건지 모르겠다
