```python
products = [
    {"name": "키보드", "price": 45000, "stock": 1, "category": "전자기기"},
    {"name": "마우스", "price": 28000, "stock": 0, "category": "전자기기"},
    {"name": "모니터", "price": 320000, "stock": 4, "category": "전자기기"},
    {"name": "책상", "price": 180000, "stock": 0, "category": "가구"},
    {"name": "의자", "price": 150000, "stock": 7, "category": "가구"},
    {"name": "키보드", "price": 45000, "stock": 1, "category": "전자기기"}
]

for_sales = []

for p in products:

    # 재고 없으면 넘어감
    if p["stock"] <= 0:
        continue

    found = False

    # 이미 같은 상품이 있는지 확인
    for sale in for_sales:
        if sale["name"] == p["name"]:
            sale["stock"] = sale["stock"] + p["stock"]
            found = True
            break

    # 같은 상품이 없으면 새로 추가
    if found == False:
        for_sales.append(p)

for sale in for_sales:
    print(
        sale["name"],
        sale["price"],
        sale["stock"]
    )
```
