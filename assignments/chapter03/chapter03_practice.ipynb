• 판매상태가 판매중이 아니거나 재고가 0 이하이면 판매불가

• 위 조건에 해당하지 않으면서 평점이 4.5 이상이고 재고가 5개 이상이면 추천

• 위 조건에 해당하지 않으면서 평점이 4.0 이상이면 일반

• 나머지는 관심 필요

• 모든 상품의 상품명, 평점, 재고, 판매상태, 결과를 확인할 수 있어야 합니다.

 

이번 문제에서는 이전 문제보다 힌트를 줄입니다.

```python
parts = [
    {"name": "키보드", "points": 4.8, "stock": 12, "sale_status": "판매중"},
    {"name": "마우스", "points": 4.6, "stock": 3, "sale_status": "판매중"},
    {"name": "모니터", "points": 4.2, "stock": 0, "sale_status": "판매중"},
    {"name": "헤드셋", "points": 4.9, "stock": 8, "sale_status": "판매중지"},
    {"name": "웹캠", "points": 3.8, "stock": 10, "sale_status": "판매중"}
]

for part in parts:
    name = part["name"]
    points = part["points"]
    stock = part["stock"]
    sale_status = part["sale_status"]

    if sale_status != "판매중" or stock <= 0:
        result = "판매불가"
        part["result"] = result
    elif points >= 4.5 and stock >= 5:
        result = "추천"
        part["result"] = result
    elif points >= 4.0:
        result = "일반"
        part["result"] = result
    else:
        result = "관심 필요"
        part["result"] = result

    print(f"상품명 {name},  평점 {points}")
    print(f"재고 {stock}, 판매상태 {sale_status}") 
    print(f"result {result}")
```
