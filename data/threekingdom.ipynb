```python
warlords = [
    {"name": "조조", "manpower": 62, "legitimacy": 63},
    {"name": "원소", "manpower": 97, "legitimacy": 21},
    {"name": "유비", "manpower": 7, "legitimacy": 55},
    {"name": "손책", "manpower": 25, "legitimacy": 41},
    {"name": "유표", "manpower": 39, "legitimacy": 25},
    {"name": "유장", "manpower": 29, "legitimacy": 43}
]

for warlord in warlords:
    name = warlord["name"]
    manpower = warlord["manpower"]
    legitimacy = warlord["legitimacy"]

    if manpower < 5 or legitimacy < 30:
        status = "역적"
    elif manpower >= 50 and legitimacy >=50:
        status = "천명후보자"
    elif manpower >= 25 and legitimacy >=40:
        status = "주요군벌"
    elif manpower >= 10 and legitimacy >=30:
        status = "지방군벌"
    else:
        status = "군소세력"
    print(
        f"{name}: 군사력 {manpower},"
        f" 정통성 {legitimacy}",
        f" 인식  {status}"
)

```
