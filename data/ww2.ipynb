```python
import pandas as pd

df = pd.read_csv("data/raw_data/countries.csv")


def classify_tier(population, strength):
    if population < 2500000 or strength < 800:
        return "tiny power"

    elif population >= 50000000 and strength >= 10000:
        return "Great power"

    elif population >= 30000000 and strength >= 6000:
        return "major power"

    elif population >= 18000000 and strength >= 3000:
        return "middle power"

    elif population >= 9000000 and strength >= 1500:
        return "regional power"

    else:
        return "minor power"


for _, country in df.iterrows():
    name = country["name"]
    population = country["population"]
    strength = country["strength"]

    tier = classify_tier(population, strength)

    print(
        f"{name}: "
        f"인구 {population:,}, "
        f"국력 {strength:,}, "
        f"티어 {tier}"
    )
```

```python
df["tier"] = df.apply(
    lambda row: classify_tier(row["population"], row["strength"]),
    axis=1
)

tier_order = [
    "Great power",
    "major power",
    "middle power",
    "regional power",
    "minor power",
    "tiny power"
]

for tier in tier_order:
    print(f"\n===== {tier} =====")

    result = df[df["tier"] == tier].sort_values(
        "strength",
        ascending=False
    )

    display(result[["name", "population", "strength"]])
```
